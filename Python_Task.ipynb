```python
import pandas as pd
import numpy as np
```

```python
#LOAD DATA
```

```python
df=pd.read_csv('Workforce_data.csv')
```

```python
print(df.shape)
print(df.head(3))
```

```python
#Cleaning with pandas and numpy
#check null
```

```python
print(df.isnull().sum())
```

```python
#pandas fillna
```

```python
df['department']=df['department'].fillna('ECE')
```

```python
#creating new column
```

```python
df['JoiningDate']=pd.to_datetime(df['JoiningDate'])
```

```python
#performance category
```

```python
df['Perf_category']=np.where(df['PerformanceScore']>=4),'High',
np.where(df['PerformanceScore']>=2,'Medium','Low')
```

```python
#salary greater than 10K
```

```python
df['salary']=np.where(df['empSalary']>=10000)
```

```python
#mean of the age
```

```python
df['age'].mean()
```
