```python
!mysql -u root -D springboard -e "select * from (select e.EmpName,e.Salary,d.DeptName,rank() over(partition by d.DeptName order by salary desc) as rnk from employee e join department d on e.EmpId=d.DeptId)x where x.rnk<3;"
```
